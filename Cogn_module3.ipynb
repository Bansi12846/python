```python
print("\n===== FOR LOOP WITH RANGE =====")
squares = ["red", "yellow", "green", "purple", "blue"]
for i in range(0, 5):
    squares[i] = "white"
print(squares)
```

```python
print("\n===== FOR LOOP OVER LIST =====")
squares = ["red", "yellow", "green"]
for square in squares:
    print(square)
```

```python
print("\n===== ENUMERATE =====")
squares = ["red", "yellow", "green"]
for i, square in enumerate(squares):
    print(i, square)
```

```python
print("\n===== WHILE LOOP =====")
squares = ['orange', 'orange', 'purple', 'blue']
Newsquares = []
i = 0
while i < len(squares) and squares[i] == 'orange':
    Newsquares.append(squares[i])
    i = i + 1
print(Newsquares)
```

```python
print("\n===== SIMPLE FUNCTION WITH DOCSTRING =====")
def function(a):
    """add 1 to a"""
    b = a + 1
    print(a, "+1 =", b)
    return b

function(5)
```

```python
print("\n===== f1 AND f2 =====")
def f1(input):
    """add 1 to input"""
    output = input + 1
    return output

def f2(input):
    """multiply input by 2"""
    output = input * 2
    return output

print("f1(3) =", f1(3))
print("f2(3) =", f2(3))
```

```python
print("\n===== BUILT-IN: len =====")
album_ratings = [10.0, 8.5, 9.5, 7.0, 7.0, 9.5, 9.0, 9.5]
L = len(album_ratings)
print("len :", L)

print("\n===== BUILT-IN: sum =====")
S = sum(album_ratings)
print("sum :", S)

print("\n===== SORTED vs SORT =====")
album_ratings = [10.0, 8.5, 9.5, 7.0, 7.0, 9.5, 9.0, 9.5]
sorted_album_rating = sorted(album_ratings)
print("sorted_album_rating :", sorted_album_rating)
print("album_ratings (unchanged) :", album_ratings)

album_ratings.sort()
print("after .sort(), album_ratings :", album_ratings)
```

```python
print("\n===== DEFINING A FUNCTION: add1 =====")
def add1(a):
    b = a + 1
    return b

print("add1(5) =", add1(5))
c = add1(10)
print("c = add1(10) ->", c)

c = add1(5)
print("c =", c)
d = add1(8)
print("d =", d)
```

```python
print("\n===== SCOPE: GLOBAL + PARAMETER =====")
def AddDC(x):               # x is a parameter here (local to the function)
    x = x + "DC"
    print(x)
    return x

x = "AC"
z = AddDC(x)
print("global x still:", x)
print("z =", z)
```

```python
print("\n===== SCOPE: LOCAL VARIABLES =====")
def Thriller():
    Date = 1982
    return Date

print("Thriller() =", Thriller())
# print(Date) # <- would give NameError: Date is local to Thriller

Date = 2017 # global Date
print("Thriller() =", Thriller()) # 1982 (local)
print("Date =", Date) # 2017 (global)

print("\n===== SCOPE: GLOBAL VARIABLE USED INSIDE FUNCTION =====")
def ACDC(y):
    print(Rating)
    return Rating + y

Rating = 9
Z = ACDC(1)
print("Z =", Z)
print("Rating =", Rating)
```

```python
print("\n===== OBJECTS: type() =====")
print(type([1, 34, 3]))
print(type('The cat is yellow'))
print(type(1))
print(type({"dog": 1, "Cat": 2}))

print("\n===== METHODS: sort() and reverse() =====")
Ratings = [10, 9, 6, 5, 10, 8, 9, 6, 2]
Ratings.sort()
print("after sort()   :", Ratings)
Ratings.reverse()
print("after reverse():", Ratings)

print("\n===== CLASS: Circle =====")
class Circle(object):
    def __init__(self, radius=3, color='red'):
        self.radius = radius
        self.color = color

    def add_radius(self, r):
        self.radius = self.radius + r
        return self.radius

    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()

print("\n===== CLASS: Rectangle =====")
class Rectangle(object):
    def __init__(self, width=2, height=3, color='r'):
        self.height = height
        self.width = width
        self.color = color

    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height, fc=self.color))
        plt.axis('scaled')
        plt.show()
print("\n===== CREATING INSTANCES: Circle =====")
RedCircle = Circle(10, "red")
print("RedCircle.radius :", RedCircle.radius)
print("RedCircle.color :", RedCircle.color)

C1 = Circle(10, "red")
print("C1.radius :", C1.radius)
print("C1.color :", C1.color)

print("\n===== CHANGING AN ATTRIBUTE =====")
C1.color = "blue"
print("C1.color :", C1.color)

print("\n===== METHOD: add_radius =====")
C1 = Circle(2, 'red')
print("before :", C1.radius)
print("C1.add_radius(8) returns :", C1.add_radius(8))
print("after :", C1.radius)

print("\n===== RED CIRCLE (radius 3) + DRAW =====")
RedCircle = Circle(3, 'red')
print("radius :", RedCircle.radius)
print("color :", RedCircle.color)
RedCircle.drawCircle()

print("\n===== BLUE CIRCLE (radius 10) + DRAW =====")
BlueCircle = Circle(10, 'blue')
print("radius :", BlueCircle.radius)
print("color :", BlueCircle.color)
BlueCircle.drawCircle()

print("\n===== RECTANGLE INSTANCES (slide 16) =====")
Object1 = Rectangle(2, 2, 'blue')
Object2 = Rectangle(3, 1, 'yellow')
print("Object1 -> height:", Object1.height, "width:", Object1.width, "color:", Object1.color)
print("Object2 -> height:", Object2.height, "width:", Object2.width, "color:", Object2.color)

print("\n===== FatYellowRectangle =====")
FatYellowRectangle = Rectangle(20, 5, 'yellow')
print("height :", FatYellowRectangle.height)
print("width :", FatYellowRectangle.width)
print("color :", FatYellowRectangle.color)
FatYellowRectangle.drawRectangle()
```
