```python
u = np.array([1, 0])
v = np.array([0, 1])

z = u + v

print(z)
```

```python
u = [1, 0]
v = [0, 1]

z = []

for n, m in zip(u, v):
    z.append(n + m)

print(z)
```

```python
u = [1, 0]
v = [0, 1]

z = []

for n, m in zip(u, v):
    z.append(n - m)

print(z)
```
