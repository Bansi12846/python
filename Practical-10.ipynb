```python
#Demonstrate NumPy array creation, indexing, slicing, reshaping, broadcasting and mathematical operations on multidimensionalarrays.
```

```python
import numpy as np
```

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print("Array:")
print(a)
```

```python
print("Element at row 1, column 2:", a[0, 1])
```

```python
print("First row:", a[0, :])
print("First two columns:")
print(a[:, 0:2])
```

```python
b = np.array([1, 2, 3, 4, 5, 6])
c = b.reshape(2, 3)

print("Reshaped array:")
print(c)
```

```python
print("Array + 10:")
print(a + 10)
```

```python
print("Addition:")
print(a + 2)
```

```python
print("Multiplication:")
print(a * 2)
```

```python
print("Square:")
print(a ** 2)
```

```python
print("Sum:", np.sum(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
```
