```python
import pandas as pd
```

```python
s = pd.Series([10, 20, 30, 40])
print("Series:")
print(s)
```

```python
data = {
    "Name": ["Bansi", "Zeny", "Pal", "Zarna"],
    "Marks": [85, 90, 75, 95],
    "City": ["Surat", "Surat", "Ahmedabad", "Surat"]
}

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)
```

```python
print("\nSelect Name column:")
print(df["Name"])

print("\nSelect first row:")
print(df.iloc[0])
```

```python
print("\nStudents with marks greater than 80:")
print(df[df["Marks"] > 80])
```

```python
print("\nGroup by City:")
print(df.groupby("City")["Marks"].mean())
```

```python
print("\nTotal Marks:", df["Marks"].sum())
print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
```

```python
print(df["Marks"].agg(["sum", "mean", "max", "min", "count"]))
```
