```python
#Use SciPy's stats module to compute statistical measures andperform statistical tests on datasets.
```

```python
from scipy import stats
```

```python
data = [10, 20, 20, 30, 40, 50]
print("Mean:",round(stats.tmean(data),4))
print("Median:", stats.scoreatpercentile(data, 50))
print("Mode:", stats.mode(data, keepdims=True).mode[0])
print("Standard Deviation:", round(stats.tstd(data),4))
print("Variance:", stats.tvar(data))
```

```python
mean = stats.tmean(data)
std = stats.tstd(data)
x = 30
probability = stats.norm.pdf(x, mean, std)
print("Normal Distribution Probability:", probability)
```

```python
t, p = stats.ttest_1samp(data, 25)

print("\nT-test")
print("T-value:", round(t,4))
print("P-value:",round(p,4))
```

```python
if p < 0.05:
    print("Significant difference")
else:
    print("No significant difference")
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

result = stats.pearsonr(x, y)

print("\nPearson Correlation Test")
print("Correlation:", result.statistic)
print("P-value:", result.pvalue)
```
