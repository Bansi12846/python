```python
#Compute and visualize Binomial and Poisson probability distributions using SciPy and Matplotlib.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson
```

```python
# Binomial Distribution
n = 15
p = 0.9

x = np.arange(0, 16)
y = binom.pmf(x, n, p)

plt.bar(x, y)
plt.title("Binomial Distribution")
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.show()
```

```python
lam = 6

x = np.arange(0, 16)
y = poisson.pmf(x, lam)

plt.bar(x, y)
plt.title("Poisson Distribution")
plt.xlabel("Number of events")
plt.ylabel("Probability")
plt.show()
```
