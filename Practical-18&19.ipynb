```python
import pandas as pd
import numpy as np

columns = ["age", "workclass", "fnlwgt", "education", "education_num",
           "marital_status", "occupation", "relationship", "race", "sex",
           "capital_gain", "capital_loss", "hours_per_week",
           "native_country", "income"]

df = pd.read_csv("adult.data", names=columns, skipinitialspace=True,
                 na_values="?")

print(df.head())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
for col in ["workclass", "occupation", "native_country"]:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Handling:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Records:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate Records After Removal:", df.duplicated().sum())
```

```python
Q1 = df["hours_per_week"].quantile(0.25)
Q3 = df["hours_per_week"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["hours_per_week"] < lower) |
              (df["hours_per_week"] > upper)]

print("\nNumber of Outliers:", len(outliers))
```

```python
df["hours_per_week"] = df["hours_per_week"].clip(lower, upper)

print("\nOutliers Treated Successfully.")
print(df["hours_per_week"].describe())
```
