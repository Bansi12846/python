```python
#List
names=["Bansi","Rensi","Zeny"]
names.append("Krupa")
print("after append:",names)
names.remove("Bansi")
print("after remove:",names)
print("index no. :",names.index("Rensi"))
names.insert(2,"Esha")
print("after insert:",names)
names.sort()
print("after sorting:",names)
names.reverse()
print("after reverse:",names)
```

```python
#Tuple
numbers = (10, 20, 30, 20, 40)
print("Tuple:", numbers)
print("Count of 20:", numbers.count(20))
print("Index of 30:", numbers.index(30))
```

```python
# Set
numbers = {10, 20, 30, 40}
print("Set:", numbers)
numbers.add(50)
print("After add:", numbers)
numbers.remove(20)
print("After remove:", numbers)
numbers.update({60, 70})
print("After update:", numbers)
```

```python
#Dictionary
student = {"name": "Bansi","age": 20,"branch": "CE"}
student["city"] = "Surat"
print("After adding city:", student)
student["age"] = 21
print("After updating age:", student)
print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
```

```python
#String
str = "python programming"
print("Uppercase:", str.upper())
print("Lowercase:", str.lower())
print("Capitalized:", str.capitalize())
print("Title:", str.title())
print("Replace:", str.replace("python", "Java"))
print("Find 'programming':", str.find("programming"))
print("Length:", len(str))
```
