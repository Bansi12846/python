```python
#if
age = 20
if age >= 18:
    print("You are eligible to vote")
```

```python
#if-else
num=int(input("enter the value of a:"))
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
#elif
marks=int(input("enter the value of a:"))
if marks >= 80:
    print("Grade A")
elif marks < 80 and marks >= 60:
    print("Grade B")
elif marks < 60 and marks >=33:
    print("Grade c")
else:
    print("Fail")
```

```python
#while Loop
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
#for Loop
for i in range(6):
    print(i)
```

```python
#Nested Loop
for i in range(2):
    for j in range(3):
        print(i, j)
```

```python
#break
for i in range(6):
    if i == 3:
        break
    print(i)
```

```python
#continue
for i in range(6):
    if i == 3:
        continue
    print(i)
```

```python
#pass
for i in range(6):
    if i == 2:
        pass
    print(i)
```
