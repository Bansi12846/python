```python
#No Argument, No Return Value
def greet():
    print("Hello world")
greet()
```

```python
#Argument, No Return Value
def greet(name):
    print("Hello", name)
greet("Bansi")
```

```python
#No Argument, With Return Value
def add():
    a = 6
    b = 4
    return a + b
sum = add()
print("Addition:", sum)
```

```python
#Argument, With Return Value
def add(a, b):
    return a + b

sum = add(6, 4)
print("Addition:", sum)
```

```python
#Default argument
def greet(name="Bansi"):
    print("Hello", name)
greet()
```

```python
import math
print("Square root:", math.sqrt(25))
print("Power:", math.pow(2, 3))
```

```python
import random
print("Random number:", random.randint(1, 10))
```

```python
import statistics
numbers = [10, 20, 30, 40, 50]
print("Mean:", statistics.mean(numbers))
print("Median:", statistics.median(numbers))
```
