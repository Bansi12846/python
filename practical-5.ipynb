```python
#text file
# Write operation
file = open("sample.txt", "w")
file.write("Hello, World!\n")
file.write("Welcome\n")
file.close()

# Read operation
file = open("sample.txt", "r")
content = file.read()
print("After writing File Content:")
print(content)
file.close()
```

```python

# Append operation
file = open("sample.txt", "a")
file.write("This line is added using append operation.\n")
file.close()

# Read again to show appended content
file = open("sample.txt", "r")
print("After Append:")
print(file.read())
file.close()
```

```python
#csv file
import csv
data=[["Roll No", "Name", "Marks"],[1, "Bansi", 85],[2, "Zeny", 95],[3, "Zarna", 92]]

# Write operation
file = open("students.csv", "w", newline="")
writer = csv.writer(file)
writer.writerows(data)
file.close()

# Read operation
file = open("students.csv", "r")
reader = csv.reader(file)
print("CSV File Content:")
for row in reader:
    print(row)
file.close()
```

```python
# Append operation
file = open("students.csv", "a", newline="")
writer = csv.writer(file)
writer.writerow([4, "Rensi", 88])
file.close()
print("\nAfter Append:")

# Read again
file = open("students.csv", "r")
reader = csv.reader(file)
for row in reader:
    print(row)
file.close()
```

```python
#binary file
# Write operation
file = open("sample.bin", "wb")
file.write( b"Hello, World!")
file.close()

# Read operation
file = open("sample.bin", "rb")
print("Binary File Content:",file.read() )
file.close()
```

```python
# Append operation
file = open("sample.bin", "ab")
file.write(b" Welcome to Python.")
file.close()

# Read again after append
file = open("sample.bin", "rb")
print("After Append:", file.read())
file.close()
```
