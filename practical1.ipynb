```python
#arithmatic operator
a=input("enter the value of a:")
a=int(a)
b=input("enter the value of b:")
b=int(b)
c=a+b
d=a-b
e=a/b
f=a*b
g=a%b
h=a**b
i=a//b
print("addition is:",c)
print("subtraction is:",d)
print("division is:",e)
print("multiplication is:",f)
print("modulo is:",g)
print("power is",h)
print("Floor division is:",i)
```

```python
#assignment operator
a = 10
a += 5
print("+= :", a)
a -= 3
print("-= :", a)
a *= 2
print("*= :", a)
a /= 4
print("/= :", a)
a //= 2
print("//= :", a)
a %= 3
print("%= :", a)
a **= 2
print("**= :", a)
```

```python
#comparision operator
a=6
b=4
print(a==b)
print(a>=b)
print(a<=b)
print(a>b)
print(a<b)
print(a!=b)
```

```python
#logical operator
a = 6
b = 4
print( a > 5 and b < 10)
print(a > 15 or b < 10)
print(not(a > 5))
```

```python
#Membership Operator
fruits = ["apple", "banana", "mango"]
print("apple in fruits:", "apple" in fruits)
print("orange in fruits:", "orange" in fruits)
print("orange not in fruits:", "orange" not in fruits)
```

```python
#Identity Operator
a = [1, 2, 3]
b = a
c = [1, 2, 3]
print("a is b:", a is b)
print("a is c:", a is c)
print("a is not c:", a is not c)
```

```python
#Bitwise Operators
a = 10
b = 3
print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```
